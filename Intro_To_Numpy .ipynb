## 2. Weather Data

Each temperature represents one day, in order, measured in degrees Celsius.

```python
temperatures = np.array([18, 21, 25, 17, 16,
                         23, 28, 30, 22, 19,
                         15, 24, 27, 20])

temperatures
```

### Filtering and Counting Days

```python
#Temperatures on days above 25 degrees
temperatures[temperatures > 25]
```

```python
#Number of days above 25 degrees
(temperatures > 25).sum()
```

```python
#Temperatures on days below 20 degrees
temperatures[temperatures < 20]
```

```python
#Number of days below 20 degrees
(temperatures < 20).sum()
```

```python
#Average temperature
avg_temp = temperatures.mean()

avg_temp
```

```python
#Hottest temperature
temperatures.max()
```

```python
#Coldest temperature
temperatures.min()
```

### Which Day Was Hottest or Coldest?

`.argmax()` returns the index of the highest value; `.argmin()` returns the index of the lowest. Add 1 to change a zero-based index into a day number. If values tie, these methods return the first matching index.

```python
#Hottest day number
temperatures.max() + 1
```

```python
#Number of days above the average temperature
(temperatures > avg_temp).sum()
```

## 3. Retail Sales

Rows represent 3 stores and columns represent 4 weeks.

```python
sales = np.array([
    [120, 150, 180, 200],
    [90, 110, 140, 160],
    [200, 220, 250, 300]
])

sales
```

### Calculating Across Rows and Columns

- 1: calculate across columns, giving one result per row (store).
- 0: calculate down rows, giving one result per column (week).
- No axis: calculate over the entire array.

The same axis rule applies to .mean()

```python
#Total sales per store - sum of each row
store_totals = sales.sum(1)

store_totals
```

```python
#Total sales per week - sum of each column
week_totals = sales.sum(0)

week_totals
```

```python
#Overall sales
sales.sum()
```

### Best Performing Store and Week

Use `.max()` on the totals to find the largest total. Add 1 to display the store or week number.

```python
#Best performing store based on total sales
store_totals.max() + 1
```

```python
#Best performing week across all stores
week_totals.max() + 1
```

```python
#Average weekly sales per store - mean of each row
sales.mean(1)
```
